```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/seasonal_agriculture_performance_dataset (3).csv')
```

## Data Cleaning and Preparation

Based on the `df.info()` output, we have identified missing values in `Rainfall_mm`, `Soil_Moisture_pct`, and `Yield_Tonnes_Ha`. Let's first quantify these missing values to understand their extent.

```python
# Display the count of missing values for each column
display(df.isnull().sum())
```

### Handling Missing Values

We have identified missing values in `Rainfall_mm` (48), `Soil_Moisture_pct` (40), and `Yield_Tonnes_Ha` (32). As these are numerical columns and the number of missing values is small, we will impute them with the mean of their respective columns.

```python
# Impute missing values with the mean
df['Rainfall_mm'].fillna(df['Rainfall_mm'].mean(), inplace=True)
df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].mean(), inplace=True)
df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].mean(), inplace=True)

# Verify that there are no more missing values
display(df.isnull().sum())
```

## Examining Agricultural Performance Across Seasons

To analyze how agricultural performance varies across seasons, we will focus on 'Profit_per_Hectare' as a key performance indicator. First, let's calculate this metric.

```python
# Calculate 'Profit_per_Hectare'
df['Profit_per_Hectare'] = df['Profit_INR'] / df['Farm_Area_Hectares']

# Display the first few rows with the new column
display(df[['Season', 'Profit_INR', 'Farm_Area_Hectares', 'Profit_per_Hectare']].head())
```

### Visualizing Profit per Hectare by Season

Now, let's visualize the distribution of 'Profit_per_Hectare' for each season using a box plot to observe differences and trends.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.boxplot(x='Season', y='Profit_per_Hectare', data=df)
plt.title('Distribution of Profit per Hectare by Season')
plt.xlabel('Season')
plt.ylabel('Profit per Hectare (INR/Hectare)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Average Profit per Hectare by Season

To further understand the seasonal performance, let's calculate the average 'Profit_per_Hectare' for each season.

```python
# Calculate and display the mean 'Profit_per_Hectare' for each season
seasonal_profit_mean = df.groupby('Season')['Profit_per_Hectare'].mean().sort_values(ascending=False)
display(seasonal_profit_mean)

# Identify important seasonal patterns and trends - Yield_Tonnes_Ha
plt.figure(figsize=(10, 6))
sns.boxplot(x='Season', y='Yield_Tonnes_Ha', data=df)
plt.title('Distribution of Yield Tonnes per Hectare by Season')
plt.xlabel('Season')
plt.ylabel('Yield Tonnes per Hectare')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Average Yield per Hectare by Season

Let's also look at the average 'Yield_Tonnes_Ha' for each season to understand productivity patterns.

```python
# Calculate and display the mean 'Yield_Tonnes_Ha' for each season
seasonal_yield_mean = df.groupby('Season')['Yield_Tonnes_Ha'].mean().sort_values(ascending=False)
display(seasonal_yield_mean)
```

## Comparing Relevant Groups: Top Crops by Season

To identify significant differences and patterns, let's examine the performance of different crops within each season. We will first find the most frequent crops in each season and then analyze their average profit per hectare.

```python
# Get the top 5 most common crops for each season
top_crops_by_season = df.groupby('Season')['Crop'].value_counts().groupby(level=0).head(5).reset_index(name='count')
display(top_crops_by_season)

# Calculate average Profit_per_Hectare for each Crop within each Season
crop_seasonal_profit = df.groupby(['Season', 'Crop'])['Profit_per_Hectare'].mean().reset_index()

# Merge with top crops to focus on the most relevant ones
top_crop_seasonal_profit = pd.merge(top_crops_by_season, crop_seasonal_profit, on=['Season', 'Crop'])

# Sort for better readability
top_crop_seasonal_profit_sorted = top_crop_seasonal_profit.sort_values(by=['Season', 'Profit_per_Hectare'], ascending=[True, False])
display(top_crop_seasonal_profit_sorted)
```

### Visualizing Top Crop Performance by Season

Let's visualize the average profit per hectare for the top crops in each season.

```python
plt.figure(figsize=(12, 8))
sns.barplot(x='Crop', y='Profit_per_Hectare', hue='Season', data=top_crop_seasonal_profit_sorted)
plt.title('Average Profit per Hectare for Top Crops by Season')
plt.xlabel('Crop')
plt.ylabel('Average Profit per Hectare (INR/Hectare)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## Investigating Relationships Between Seasonal Conditions and Agricultural Outcomes

To understand why profit and yield vary across seasons, let's examine the distribution of key environmental conditions (Rainfall, Average Temperature, Humidity, Sunlight Hours) for each season. This will help us identify patterns that might correlate with agricultural outcomes.

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('Distribution of Environmental Conditions by Season', fontsize=16)

sns.boxplot(x='Season', y='Rainfall_mm', data=df, ax=axes[0, 0])
axes[0, 0].set_title('Rainfall (mm) by Season')
axes[0, 0].set_xlabel('Season')
axes[0, 0].set_ylabel('Rainfall (mm)')

sns.boxplot(x='Season', y='Avg_Temperature_C', data=df, ax=axes[0, 1])
axes[0, 1].set_title('Average Temperature (°C) by Season')
axes[0, 1].set_xlabel('Season')
axes[0, 1].set_ylabel('Average Temperature (°C)')

sns.boxplot(x='Season', y='Humidity_pct', data=df, ax=axes[1, 0])
axes[1, 0].set_title('Humidity (%) by Season')
axes[1, 0].set_xlabel('Season')
axes[1, 0].set_ylabel('Humidity (%)')

sns.boxplot(x='Season', y='Sunlight_Hours_Day', data=df, ax=axes[1, 1])
axes[1, 1].set_title('Sunlight Hours per Day by Season')
axes[1, 1].set_xlabel('Season')
axes[1, 1].set_ylabel('Sunlight Hours per Day')

plt.tight_layout(rect=[0, 0.03, 1, 0.96])
plt.show()
```
